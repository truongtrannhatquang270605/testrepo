```python
# **1.**  Xây dựng một lớp đối tượng `Student` gồm các phương thức:  
# `__init__(id, name)`  
# `set_id(id)`  *# Thiết lập id cho sinh viên*  
# `get_id()`  *# Trả về id của sinh viên*  
# `set_name(name)`  *# Thiết lập tên của sinh viên*  
# `get_name()`  *# Trả về tên của sinh viên*  
# `__str__()  `*# Trả về một chuỗi gồm các thông tin id và name của sinh viên*  
  
# Kiểm thử bằng cách tạo sinh viên và xuất ra dùng lệnh print:  
# `stu01 = Student([MSSV của SV], ‘[Họ và Tên của SV]’)`   
# `print(stu01)`  
   
# `stu02 = Student()`   
# `stu02.set_id([MSSV của SV])`  
# `stu02.set_name(‘[Họ và Tên của SV]’)`  
# `print(stu02.get_id(),'-',stu02.get_name())`  
class Student:
    def __init__(self,id=0,name="khong co"):
        self._id=id
        self._name=name
    def set_id(self,id):
        self._id=id
    def get_id(self):
        return self._id
    def set_name(self,name):
        self._name=name
    def get_name(self):
        return self._name
    def __str__(self):
        return f"id: {self._id}, name: {self._name}"
stu=Student(1,"nguyen van a")
print(stu)
stu1=Student()
stu1.set_id(2)
stu1.set_name("tran thi buoi")
print("id:",stu1.get_id(),"name:",stu1.get_name())
```

```python
# **2.** Xây dựng một lớp đối tượng $StudentScore$, kế thừa từ lớp $Student$, cập nhật và bổ sung các thuộc tính và phương thức như sau:  
# `__init__(id, name, regular, midterm, finalExam, courseCode)`   
# $\hspace{15pt}$`self._summaryScore = None`                                 *# Điểm tổng kết*   
# `enrollment(id, name, courseCode)`  *# Nhập thông tin sinh viên vào lớp học phần*  
# `set_score(regular, midterm, finalExam)`  *# Nhập điểm thường kỳ, giữa kỳ và cuối kỳ*  
# `summaryScore()`  *# Tính điểm tổng kết summaryScore = 20% regular + 30% midterm + 50% FinalExam*   
# `__str__()` *# Trả về các thông tin gồm id, tên, các cột điểm và mã học phần của sinh viên*  
class StudentScore(Student):
    def __init__(self, id=0, name="khong co",regular=0,midterm=0,finalExam=0,courseCode="A0"):
        super().__init__(id, name)
        self._regular = regular
        self._midterm = midterm
        self._finalExam= finalExam
        self._courseCode=courseCode
        self._summaryScore=None
    def enrollment(self, id=0, name="khong co", courseCode="A0"):
        self._id=id
        self._name=name
        self._courseCode=courseCode
    def set_score(self,regular=0,midterm=0,finalExam=0):
        self._regular=regular
        self._midterm=midterm
        self._finalExam=finalExam
    def summaryScore (self):
        return 0.2*self._regular+0.3*self._midterm+0.5*self._finalExam
    def __str__(self):
        return f"id:{self._id},name: {self._name},regular: {self._regular},midterm: {self._midterm},finalExam: {self._finalExam},courseCode:{self._courseCode}, summaryScore: {self.summaryScore()}"
s1=StudentScore(22,"nguyen van trung",9,10,10,"A1")
print(s1)
```

```python
# **3.** Xây dựng một lớp đối tượng `courseTranscript` để lưu trữ bảng điểm của lớp học phần. Sức chứa của danh sách sẽ tự động mở rộng thêm 5 chỗ mỗi khi đầy.   
# `__init__(capacity = 5)`  *# Tạo một mảng rỗng*    
# $\hspace{15pt}$`self._num = 0`                                 *# Số sinh viên trong lớp*   
# $\hspace{15pt}$`self._capacity = capacity`                          *# Sức chứa của danh sách*     
# $\hspace{15pt}$`self._list = self._make_array(self._capacity)`  *# low-level array*    
# `__len__()`  *# Trả về số sinh viên được lưu trữ trong bảng*    
# `__getitem__(pos)`  *# Trả về sinh viên tại vị trí pos*    
# `_make_array(cap)`    *# Trả về mảng mới với sức chứa cap*  
# `addStudent(studentScore student)`  *# Cập nhật sinh viên vào danh sách*  
# `resize(cap)`  *# Cập nhật sức chứa của danh sách*   
# `find(id)`  *# Tìm thông tin sinh viên theo ID, nếu có thì in ra, nếu không thì báo không tìm thấy*   
# `sumScore()`  *# Tính điểm tổng kết cho toàn bộ sinh viên có trong danh sách*  
# `sortedList()`  *# Sắp xếp lại danh sách theo thứ hạng điểm tổng kết từ cao đến thấp, tính lại điểm tổng kết cho toàn bộ sinh viên trước khi sắp xếp*   
# `top(num)`  *# Liệt kê ra top điểm cao nhất với số lượng $num$ *   
# `__str__(self)`  *# Trả về danh sách sinh viên gồm id, name và summaryScore *   
import ctypes
class CourseTranscript:
    def __init__(self,capacity=5):
        self._num=0
        self._capacity=capacity
        self._list=self._make_array(self._capacity)
    def __len__(self):
        return self._num
    def __getitem__(self,pos):
        if 0<=pos<self._num:
            return self._list[pos]
        raise IndexError("loi chi muc")
    def _make_array(self,cap):
        return (ctypes.py_object*cap)()
    def addStudent(self,student):
        if self._num==self._capacity:
            self.resize(self._capacity+5)
        self._list[self._num]=student
        self._num+=1
    def resize(self, cap):
        new_list= [None]*cap
        for i in range (self._num):
            new_list[i]=self._list
        self._list=new_list
        self._capacity=cap
    def find(self, id):
        for student in self._list[:self._num]:
            if student and student.get_id()==id:
                return str(student)
        return ("could not found")
    def sumScore(self):
        for student in self._list[:self._num]:
            if student is not None:
                student._summaryScore=student.summaryScore()
    def sortedList(self):
        self.sumScore()
        sorted_list= sorted(self._list[:self._num],key=lambda s: s._summaryScore,reverse=True)
        for i in range(self._num):
            self._list[i]=sorted_list[i]
    def top(self, num):
        self.sortedList()
        return [str(self._list[i]) for i in range (min(num,self._num))]  
    def __str__(self):
        self.sortedList()
        return "\n".join([str(self._list[i])for i in range (self._num)])
s3=StudentScore(4,"tran thi n",8,9,10,"A3")
s4=StudentScore(5,"tran thi 8",1,5,9,"A3")
s5=StudentScore(6,"tran van tai",0,9,10,"A3")
tran=CourseTranscript()

tran.addStudent(s3)
tran.addStudent(s4)
tran.addStudent(s5)
print(tran)
print("========================")
print("\n".join(tran.top(2)))       
```

```python
# **4.** Tạo một cấu trúc hàng đợi $reportQueue$ theo nguyên lý xoay vòng, có sức chứa cố định là 3, 
# để lưu trữ thứ tự các sinh viên đăng ký báo cáo tiểu luận, theo nguyên tắc ai đăng ký trước thì được báo cáo trước. 
# Đưa ra các tình huống kiểm thử để thử nghiệm các tính năng của hàng đợi này, đặc biệt là khả năng xoay vòng.
class ReportQueue:
    succhuacodinh=3
    def __init__(self):
        self._dulieu=[None]*ReportQueue.succhuacodinh
        self._soluong=0
        self._dauhangdoi=0
    def __len__(self):
        return self._soluong
    def larong(self):
        return self._soluong==0
    def laday(self):
        return self.laday==ReportQueue.succhuacodinh
    def dautien(self):
        if self.larong():
            return "hang doi dang rong"
        return self._dulieu[self._dauhangdoi]
    def xoaphantu(self):
        if self.larong():
            return "hang doi rong"
        ketqua=self._dulieu[self._dauhangdoi]
        self._dulieu[self._dauhangdoi]=None
        self._dauhangdoi=(self._dauhangdoi+1)%ReportQueue.succhuacodinh
        self._soluong-=1
        return ketqua
    def themphantu(self,e):
        if self.laday():
            return "hang doi day"
        vitri= (self._dauhangdoi+self._soluong)%ReportQueue.succhuacodinh
        self._dulieu[vitri]=e
        self._soluong+=1
    def __str__(self):
        return str([self._dulieu[(self._dauhangdoi+i) % ReportQueue.succhuacodinh] for i in range (self._soluong)])
hangdoi = ReportQueue()
hangdoi.themphantu("s1") 
hangdoi.themphantu("s2") 
hangdoi.themphantu("s3") 
print(hangdoi)
print(hangdoi.xoaphantu())
print(hangdoi)
hangdoi.themphantu("s4") 
print(hangdoi)
```
